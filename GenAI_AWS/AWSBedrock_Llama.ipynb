```python
import boto3
import json
from botocore.exceptions import ClientError

client = boto3.client('bedrock-runtime', region_name='us-east-1')
client
```

```python
model_id = 'meta.llama3-8b-instruct-v1:0'
```

### Using InvokeModel API to get response from the model

```python

prompt = "What is AWS Bedrock?"

#Embed the prompt in Llama3's instruction format
formatted_prompt = f"""
<|begin_of_text|><|start_header_id|>user<|end_header_id|>
{prompt}
<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>
"""

formatted_prompt
```

```python
# Format the request payload using the model's native structure.
native_request = {
    "prompt": formatted_prompt,
    "max_gen_len": 512,
    "temperature": 0.5,
}

# Convert the native request to JSON.
request = json.dumps(native_request)

request
```

```python
try:
    response = client.invoke_model(
        modelId=model_id,
        body=request
    )
    
    model_response = json.loads(response['body'].read())

    response_text = model_response["generation"]
    print("Model Response:")
    print(response_text)

except (ClientError, Exception) as e:
    print(f"Error invoking model: {e}")
```

#### Using Converse API to get response from the model

```python
user_message = "What is AWS Lambda? Explain in brief."
conversation = [
    {"role": "user", "content": [{"text": user_message}]}
]

try:
    response = client.converse(
        modelId=model_id,
        messages=conversation,
        inferenceConfig={"maxTokens": 512, "temperature": 0.5, "topP": 0.9},
        system = [{"text":"Act as an AWS Tutor. Answer the user's question with analogies and examples."}]
    )
    response_text = response['output']['message']['content'][0]['text']
    print(response_text)

except (ClientError, Exception) as e:
    print(f"Error invoking model: {e}")
```
